# Чтение и первичная обработка данных

В этом ноутбуке используются два CSV-файла по теме проекта **«Нобелевские лауреаты: страны, области премии и годы награждения»**:

- `nobel_laureates.csv` — данные о людях, получивших Нобелевскую премию;
- `nobel_awards.csv` — общая таблица награждений Нобелевской премией.

Цель этапа — загрузить данные, проверить структуру таблиц, переименовать столбцы, привести даты и годы к корректным типам, проверить пропуски и посчитать базовую статистику.


## 1. Импорт библиотек

```python
# подключаем библиотеку для работы с таблицами
import pandas as pd
from IPython.display import display
```

## 2. Загрузка CSV-файлов из GitHub

```python
# указываем владельца и название репозитория
repo_owner = "Nikto555"
repo_name = "python-ai-Sorokin-Kirill"

# собираем базовую ссылку на папку data
base_url = f"https://raw.githubusercontent.com/{repo_owner}/{repo_name}/main/data"

# указываем raw-ссылки на csv-файлы
laureates_url = f"{base_url}/nobel_laureates.csv"
awards_url = f"{base_url}/nobel_awards.csv"

# читаем csv-файлы
df_laureates = pd.read_csv(laureates_url)
df_awards = pd.read_csv(awards_url)

# выводим размеры таблиц
print(f"nobel_laureates.csv: {df_laureates.shape[0]} строк, {df_laureates.shape[1]} столбцов")
print(f"nobel_awards.csv: {df_awards.shape[0]} строк, {df_awards.shape[1]} столбцов")

# показываем первые строки
display(df_laureates.head())
display(df_awards.head())
```

## 3. Проверка исходных столбцов

```python
# смотрим исходные названия столбцов
print("Столбцы nobel_laureates.csv:")
print(df_laureates.columns.tolist())

print("\nСтолбцы nobel_awards.csv:")
print(df_awards.columns.tolist())
```

## 4. Переименование столбцов

```python
# переименовываем столбцы в таблице лауреатов
df_laureates = df_laureates.rename(columns={
    "laureate": "laureate_url",
    "laureateLabel": "laureate",
    "prize": "prize_url",
    "prizeLabel": "prize",
    "awardDate": "award_date",
    "awardYear": "award_year",
    "citizenshipLabel": "citizenship",
    "genderLabel": "gender",
    "birthDate": "birth_date",
    "birthYear": "birth_year",
    "occupationLabel": "occupation"
})

# переименовываем столбцы в таблице награждений
df_awards = df_awards.rename(columns={
    "recipient": "recipient_url",
    "recipientLabel": "recipient",
    "recipientType": "recipient_type",
    "prize": "prize_url",
    "prizeLabel": "prize",
    "awardDate": "award_date",
    "awardYear": "award_year",
    "countryLabel": "country"
})

# проверяем результат
display(df_laureates.head())
display(df_awards.head())
```

## 5. Преобразование типов данных

```python
# преобразуем годы к числовому типу
df_laureates["award_year"] = pd.to_numeric(df_laureates["award_year"], errors="coerce")
df_laureates["birth_year"] = pd.to_numeric(df_laureates["birth_year"], errors="coerce")
df_awards["award_year"] = pd.to_numeric(df_awards["award_year"], errors="coerce")

# преобразуем даты к datetime
df_laureates["award_date"] = pd.to_datetime(df_laureates["award_date"], errors="coerce", utc=True)
df_laureates["birth_date"] = pd.to_datetime(df_laureates["birth_date"], errors="coerce", utc=True)
df_awards["award_date"] = pd.to_datetime(df_awards["award_date"], errors="coerce", utc=True)

# проверяем типы данных
print("Типы данных nobel_laureates:")
print(df_laureates.dtypes)

print("\nТипы данных nobel_awards:")
print(df_awards.dtypes)
```

## 6. Проверка пропусков

```python
# проверяем пропуски в таблице лауреатов
print("Пропуски в nobel_laureates:")
display(df_laureates.isna().sum().reset_index().rename(columns={"index": "column", 0: "missing_count"}))

# проверяем пропуски в таблице награждений
print("\nПропуски в nobel_awards:")
display(df_awards.isna().sum().reset_index().rename(columns={"index": "column", 0: "missing_count"}))
```

## 7. Базовая статистика

```python
# базовая статистика по числовым столбцам
print("Статистика по таблице лауреатов:")
display(df_laureates[["award_year", "birth_year"]].describe())

print("\nСтатистика по таблице награждений:")
display(df_awards[["award_year"]].describe())
```

## 8. Анализ категориальных признаков

```python
# смотрим распределение по премиям
print("Топ премий в nobel_laureates:")
display(df_laureates["prize"].value_counts().head(10))

print("\nТоп стран гражданства лауреатов:")
display(df_laureates["citizenship"].value_counts().head(10))

print("\nРаспределение по полу:")
display(df_laureates["gender"].value_counts())

print("\nТипы получателей в nobel_awards:")
display(df_awards["recipient_type"].value_counts())

print("\nТоп премий в nobel_awards:")
display(df_awards["prize"].value_counts().head(10))
```

## Итоги первичной обработки данных

В этом ноутбуке были загружены два CSV-файла по теме Нобелевских премий.

Файл `nobel_laureates.csv` содержит данные о людях-лауреатах: имя лауреата, премию, дату и год награждения, страну гражданства, пол, дату рождения, год рождения и род занятий.

Файл `nobel_awards.csv` содержит общие данные о награждениях: получателя, тип получателя, премию, дату и год награждения, а также страну.

В ходе работы были проверены размеры таблиц, исходные названия столбцов, первые строки данных, пропуски и базовая статистика. Столбцы были переименованы в короткие и понятные имена, даты были преобразованы к типу datetime, а годы — к числовому типу.

Эти данные подходят для дальнейшего анализа, потому что содержат категориальные признаки, такие как лауреат, страна, пол, премия и род занятий, а также числовые признаки, связанные с годами рождения и награждения.
